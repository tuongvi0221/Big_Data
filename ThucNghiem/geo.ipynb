```python
import requests

def reverse_geocode(lat, lon):
    url = f"https://nominatim.openstreetmap.org/reverse?lat={lat}&lon={lon}&format=json"
    headers = {
        "User-Agent": "MyReverseGeocodeApp/1.0 (22521658@gm.uit.edu.vn)"
    }

    try:
        response = requests.get(url, headers=headers, timeout=10)

        if response.status_code == 403:
            return "❌ Bị chặn – hãy thử đổi mạng hoặc đợi vài phút"

        if response.status_code != 200:
            return f"❌ Lỗi HTTP {response.status_code}"

        data = response.json()
        return data.get("display_name", "⚠️ Không tìm thấy địa chỉ")

    except requests.exceptions.Timeout:
        return "⏳ Quá thời gian chờ – mạng chậm hoặc server quá tải"
    except requests.exceptions.RequestException as e:
        return f"⚠️ Lỗi kết nối: {str(e)}"

# Toạ độ ví dụ
latitude =  16.08516825
longitude = 108.2288678

# Lấy địa chỉ
address = reverse_geocode(latitude, longitude)

# Hiển thị
print(f"📍 Địa chỉ: {address}")
```

Lấy các vị trí gần tọa độ tương ứng

```python
import requests
from math import radians, cos, sin, sqrt, atan2

# Hàm tính khoảng cách giữa hai tọa độ (Haversine formula)
def haversine(lat1, lon1, lat2, lon2):
    R = 6371  # Bán kính Trái Đất (km)
    dlat = radians(lat2 - lat1)
    dlon = radians(lon2 - lon1)
    
    a = sin(dlat/2)**2 + cos(radians(lat1)) * cos(radians(lat2)) * sin(dlon/2)**2
    c = 2 * atan2(sqrt(a), sqrt(1-a))
    
    return R * c  # Khoảng cách (km)

# Hàm tìm các địa điểm xung quanh
def find_nearby_places(lat, lon, radius=3000):
    overpass_url = "http://overpass-api.de/api/interpreter"
    overpass_query = f"""
    [out:json];
    (
      node(around:{radius},{lat},{lon})["amenity"="hospital"];    // Bệnh viện
      node(around:{radius},{lat},{lon})["amenity"="school"];      // Trường học
      node(around:{radius},{lat},{lon})["amenity"="marketplace"]; // Chợ
      node(around:{radius},{lat},{lon})["shop"="mall"];           // Trung tâm thương mại
    );
    out body;
    """

    headers = {"User-Agent": "MyOSMApp/1.0 (your_email@example.com)"}
    response = requests.get(overpass_url, params={"data": overpass_query}, headers=headers)

    if response.status_code != 200:
        print("❌ Không thể truy vấn dữ liệu, thử lại sau!")
        return []

    data = response.json()
    places = []
    for element in data.get("elements", []):
        name = element.get("tags", {}).get("name", "Không có tên")
        place_lat = element["lat"]
        place_lon = element["lon"]
        distance = haversine(lat, lon, place_lat, place_lon)  # Tính khoảng cách
        places.append((name, distance))

    return sorted(places, key=lambda x: x[1])  # Sắp xếp theo khoảng cách

# Chạy hàm với tọa độ của bạn

# Thử với tọa độ đã có
latitude = 21.0361782
longitude = 105.7822685

results = find_nearby_places(latitude, longitude)
if results:
    for place in results:
        print(f"📍 {place[0]} - Khoảng cách: {place[1]:.2f} km")
else:
    print("⚠️ Không tìm thấy địa điểm nào trong phạm vi 3km.")
```

Thêm cột tọa độ vào file

```python
import pandas as pd
import requests
import time
import os

# Đọc dữ liệu từ file
df = pd.read_csv("batdongsan_with_dinh_vi.csv", encoding="utf-8-sig")

# Nếu đã có file tọa độ thì đọc để tiếp tục
if os.path.exists("batdongsan_with_coordinates.csv"):
    df_coords = pd.read_csv("batdongsan_with_coordinates.csv", encoding="utf-8-sig")
    df.update(df_coords)  # Cập nhật những hàng đã có tọa độ
else:
    df["Latitude"] = None
    df["Longitude"] = None

# Hàm lấy tọa độ từ OpenStreetMap
def get_coordinates_osm(address):
    url = f"https://nominatim.openstreetmap.org/search?q={address}&format=json"
    headers = {
        "User-Agent": "MyGeocodingApp/1.0 (22521658@gm.uit.edu.vn)"
    }
    response = requests.get(url, headers=headers)
    if response.status_code == 403:
        print("❌ Bị chặn! Hãy thử đổi mạng hoặc dùng API khác.")
        return None, None
    data = response.json()
    if data:
        return float(data[0]["lat"]), float(data[0]["lon"])
    return None, None

# Lặp qua từng dòng
for idx, row in df.iterrows():
    if pd.isna(row['Latitude']) or pd.isna(row['Longitude']):
        address = row['Định Vị']
        print(f"📍 [{idx}] Lấy tọa độ cho: {address}")
        lat, lon = get_coordinates_osm(address)
        df.at[idx, 'Latitude'] = lat
        df.at[idx, 'Longitude'] = lon
        # Lưu tạm vào file
        df.to_csv("batdongsan_with_coordinates.csv", index=False, encoding="utf-8-sig")
        time.sleep(1)  # tránh bị chặn

print("✅ Đã hoàn thành toàn bộ và lưu vào 'batdongsan_with_coordinates.csv'")
```

Xử lý tọa độ cho những định vị khó tim bâc 1

```python
import pandas as pd
import requests

def get_coordinates_from_address(address):
    """
    Lấy tọa độ từ địa chỉ nếu chưa có tọa độ trong dữ liệu.
    """
    url = f"https://nominatim.openstreetmap.org/search?q={address}&format=json"
    headers = {
        "User-Agent": "MyGeocodingApp/1.0"
    }
    
    response = requests.get(url, headers=headers)
    
    if response.status_code == 200:
        data = response.json()
        if data:
            lat = float(data[0]["lat"])
            lon = float(data[0]["lon"])
            return lat, lon
    return None, None

def extract_relevant_address(full_address):
    """
    Tách phần địa chỉ cần tìm từ một địa chỉ đầy đủ, sau dấu phẩy đầu tiên.
    """
    # Tìm vị trí dấu phẩy đầu tiên và cắt chuỗi từ đó
    comma_index = full_address.find(',')
    if comma_index != -1:
        return full_address[comma_index+1:].strip()  # Lấy phần sau dấu phẩy đầu tiên
    return full_address  # Nếu không tìm thấy dấu phẩy, trả về toàn bộ địa chỉ

# Đọc dữ liệu từ CSV
df = pd.read_csv("batdongsan_with_coordinates.csv")

# Kiểm tra các dòng chưa có tọa độ và tự động lấy tọa độ nếu thiếu
for index, row in df.iterrows():
    if pd.isna(row['Latitude']) or pd.isna(row['Longitude']):
        address = row['Định Vị']
        relevant_address = extract_relevant_address(address)
        print(f"Đang tìm tọa độ cho địa chỉ: {relevant_address}")

        # Lấy tọa độ từ OpenStreetMap
        latitude, longitude = get_coordinates_from_address(relevant_address)
        
        if latitude and longitude:
            print(f"Tọa độ: latitude = {latitude}, longitude = {longitude}")
            df.at[index, 'Latitude'] = latitude
            df.at[index, 'Longitude'] = longitude
        else:
            print("Không tìm thấy tọa độ cho địa chỉ này.")

# Lưu lại CSV với tọa độ mới (nếu có)
df.to_csv("updated_file.csv", index=False)

```

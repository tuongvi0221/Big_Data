```python
import pandas as pd

# Đọc dữ liệu
df = pd.read_csv("batdongsan.csv", encoding="utf-8-sig")

# 1. Dữ liệu có ID: loại trùng theo 'id'
df_with_id = df[df["id"].notnull()]
df_with_id = df_with_id.drop_duplicates(subset="id", keep="first")

# 2. Dữ liệu không có ID: loại trùng theo 'Tên dự án'
df_no_id = df[df["id"].isnull()]
df_no_id = df_no_id.drop_duplicates(subset="Tên dự án", keep="first")

# 3. Gộp lại
df_cleaned = pd.concat([df_with_id, df_no_id]).reset_index(drop=True)

# 4. Kiểm tra lại
print("✅ Số dòng sau khi làm sạch:", len(df_cleaned))
print("🟢 Trùng ID còn không:", df_cleaned[df_cleaned["id"].notnull()].duplicated(subset="id").any())
print("🔵 Trùng Tên dự án (nếu ID NaN) còn không:", df_cleaned[df_cleaned["id"].isnull()].duplicated(subset="Tên dự án").any())

# (Tùy chọn) Xuất file sạch ra
df_cleaned.to_csv("batdongsan_cleaned.csv", index=False, encoding="utf-8-sig")
```
